# 🎮 AI-Based Tic-Tac-Toe Game

## 📌 Project Overview

This project implements an **AI-based Tic-Tac-Toe game** in Python where a human player competes against an AI opponent.

The AI uses the **Minimax algorithm** to evaluate possible game states and select an optimal move. The game is played on a **3×3 board**, where:

* 👤 Human Player → **X**
* 🤖 AI Player → **O**

The project also includes input validation, win and draw detection, multiple rounds, and a scoreboard to track the results of the game session.

## 🎮 How the Game Works

The game follows these steps:

1. A new **3×3 Tic-Tac-Toe board** is created.
2. The human player starts as **X**.
3. The player selects an available position from **1 to 9**.
4. The game checks whether the move results in a **win or draw**.
5. The AI evaluates the available moves using the **Minimax algorithm**.
6. The AI selects the move with the best calculated score.
7. Turns continue alternately between the player and AI.
8. The game ends when a player wins or the board becomes full.
9. The scoreboard is updated.
10. The player can choose to start another round.

# 🧠 Minimax Algorithm

The **Minimax algorithm** allows the AI to evaluate possible future moves before making a decision.

The algorithm considers three terminal situations:

| Game Result   |    Score |
| ------------- | -------: |
| 🤖 AI wins    | Positive |
| 👤 Human wins | Negative |
| 🤝 Draw       |        0 |

The algorithm recursively explores possible moves using:

* **Maximizing Player:** AI (`O`)
* **Minimizing Player:** Human (`X`)

The **depth of the game tree** is also considered when calculating the score. This allows the AI to distinguish between faster and slower winning outcomes.

---

## 🔍 AI Decision Process

For every available position, the AI:

1. Temporarily places `O` on the board.
2. Calls the **Minimax** function recursively.
3. Evaluates the resulting game state.
4. Undoes the temporary move using **backtracking**.
5. Compares the scores of all available moves.
6. Selects the move with the **highest score**.

This allows the AI to make decisions based on the possible future consequences of each move.

````python
# Scoreboard tracking across matches
scores = {"Player (X)": 0, "AI (O)": 0, "Ties": 0}


def create_board():
    """Initializes a 3x3 board as a 9-element list of empty spaces."""
    return [" "] * 9


def print_board(board):
    """Renders the 3x3 grid with index numbers for open spots to guide the player."""
    display = [str(i + 1) if spot == " " else spot for i, spot in enumerate(board)]
    print("\n")
    print(f" {display[0]} | {display[1]} | {display[2]} ")
    print("---+---+---")
    print(f" {display[3]} | {display[4]} | {display[5]} ")
    print("---+---+---")
    print(f" {display[6]} | {display[7]} | {display[8]} ")
    print("\n")
def check_winner(board, player):
    """Checks rows, columns, and diagonals for 3 matching symbols."""
    win_conditions = [
        # Horizontal rows
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        # Vertical columns
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        # Diagonals
        (0, 4, 8),
        (2, 4, 6),
    ]
    for a, b, c in win_conditions:
        if board[a] == board[b] == board[c] == player:
            return True
    return False


def is_board_full(board):
    """Returns True if there are no empty spaces remaining on the board."""
    return " " not in board


def get_available_moves(board):
    """Returns a list of integer indices (0-8) for all empty spots."""
    return [i for i, spot in enumerate(board) if spot == " "]
````

````python
def minimax(board, depth, is_maximizing):
    """Recursively evaluates the game tree to calculate optimal moves."""
    # Terminal State Evaluations
    if check_winner(board, "O"):  # AI wins
        return 10 - depth
    if check_winner(board, "X"):  # Human wins
        return depth - 10
    if is_board_full(board):  # Tie
        return 0

    if is_maximizing:
        best_score = -float("inf")
        for move in get_available_moves(board):
            board[move] = "O"  # Simulate AI move
            score = minimax(board, depth + 1, False)
            board[move] = " "  # Undo move (backtrack)
            best_score = max(score, best_score)
        return best_score
    else:
        best_score = float("inf")
        for move in get_available_moves(board):
            board[move] = "X"  # Simulate human move
            score = minimax(board, depth + 1, True)
            board[move] = " "  # Undo move (backtrack)
            best_score = min(score, best_score)
        return best_score


def get_best_move(board):
    """Finds the optimal grid index for the AI ('O')."""
    best_score = -float("inf")
    best_move = None

    for move in get_available_moves(board):
        board[move] = "O"
        score = minimax(board, 0, False)
        board[move] = " "

        if score > best_score:
            best_score = score
            best_move = move

    return best_move
````

````python
def get_human_move(board):
    """Prompts the player to enter a move (1-9) and validates the input."""
    while True:
        try:
            choice = input("Enter your move (1-9): ").strip()

            # Ensure the user entered a single valid digit from 1 to 9
            if not choice.isdigit():
                print("Invalid input! Please enter a number between 1 and 9.")
                continue

            move = int(choice) - 1  # Convert 1-9 to 0-8 list index

            if move not in range(9):
                print("Out of range! Please choose a number from 1 to 9.")
                continue

            if board[move] != " ":
                print("That spot is already taken! Choose an empty spot.")
                continue

            return move

        except (ValueError, IndexError):
            print("Invalid input! Try again.")
````

````python
def play_round():
    """Runs a single round of Tic-Tac-Toe between Human and Minimax AI."""
    board = create_board()
    current_player = "X"  # Human starts first

    print("=" * 45)
    print("New Round! You are 'X' and AI is 'O'.")
    print_board(board)

    while True:
        if current_player == "X":
            move = get_human_move(board)
            board[move] = "X"
        else:
            print("AI is calculating the optimal move...")
            move = get_best_move(board)
            board[move] = "O"

        print_board(board)

        # Check win condition
        if check_winner(board, current_player):
            if current_player == "X":
                print("Congratulations! You won!")
                scores["Player (X)"] += 1
            else:
                print("AI wins! Minimax played optimally.")
                scores["AI (O)"] += 1
            break

        # Check tie condition
        if is_board_full(board):
            print("It's a draw!")
            scores["Ties"] += 1
            break

        # Switch turns
        current_player = "O" if current_player == "X" else "X"

    # Display updated scoreboard
    print("-" * 45)
    print(
        f"Scoreboard: Player (X): {scores['Player (X)']} | AI (O): {scores['AI (O)']} | Ties: {scores['Ties']}"
    )
    print("-" * 45)


def start_game():
    """Manages the full game session with multiple replay rounds."""
    while True:
        play_round()
        play_again = input("\nDo you want to play again? (y/n): ").strip().lower()
        if play_again not in ("y", "yes"):
            print("\nThanks for playing! Final Scores:")
            print(
                f"Player (X): {scores['Player (X)']} | AI (O): {scores['AI (O)']} | Ties: {scores['Ties']}"
            )
            print("Session ended.")
            break


# Start the game session
start_game()
````
